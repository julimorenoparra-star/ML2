## **Datos**

```python
# librerias
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
import os
import pandas as pd
import glob
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score, calinski_harabasz_score, davies_bouldin_score
from sklearn.metrics.pairwise import cosine_similarity



# Función para cargar el DataFrame de Spotify
def load_spotify_dataframe():
    try:
        import kagglehub
        path = kagglehub.dataset_download("maharshipandya/-spotify-tracks-dataset")
        print(f"Kaggle path: {path}")

        # Busca un archivo CSV con "track" en el nombre, o toma el primero
        csv_files = glob.glob(os.path.join(path, "*.csv"))
        csv_path = next((f for f in csv_files if "track" in os.path.basename(f).lower()), None) or (
            csv_files[0] if csv_files else None
        )

        if not csv_path:
            raise FileNotFoundError("No se encontró ningún CSV en el dataset de Kaggle.")

        return pd.read_csv(csv_path)

    except Exception as e:
        print(f"[Aviso] Kaggle no disponible o error: {e}")
        fallback = "/mnt/data/dataset.csv"
        print(f"Usando fallback local: {fallback}")
        return pd.read_csv(fallback)
    
df = load_spotify_dataframe()

df.head(10)

```

```python
# Seleccionar características de audio (como en el original)
audio_features = [
    'danceability', 'energy', 'key', 'loudness', 'mode', 'speechiness',
    'acousticness', 'instrumentalness', 'liveness', 'valence', 'tempo'
]

X = df[audio_features].copy()
X = X.dropna()
df = df.loc[X.index].copy()

```

## **Exploración de los datos**

```python

print(f" Se tienen {X.shape[0]} canciones y {len(audio_features)} características.")

# Estadísticas descriptivas (como en el original)
print("Estadísticas descriptivas:")
display(X.describe())

# Heatmap de correlación (visualización clave del original)
plt.figure(figsize=(10, 8))
sns.heatmap(X.corr(), annot=True, cmap='coolwarm', center=0, vmax=1, vmin=-1)
plt.title('Correlación entre Características de Audio')
plt.tight_layout()
plt.show()
```

## **Preprocesamiento (Estandarización y PCA)**

```python

# Estandarización 
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# PCA 
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

print(f"Varianza explicada: PC1 = {pca.explained_variance_ratio_[0]:.3%}, PC2 = {pca.explained_variance_ratio_[1]:.3%}")
print(f"Total: {pca.explained_variance_ratio_.sum():.3%}")

# Agregar a df
df_pca = pd.DataFrame(X_pca, columns=['PC1', 'PC2'], index=df.index)
df = pd.concat([df, df_pca], axis=1)

# Scatter plot inicial 
plt.figure(figsize=(10, 8))
plt.scatter(df['PC1'], df['PC2'], alpha=0.5, s=10)
plt.xlabel('PC1')
plt.ylabel('PC2')
plt.title('Datos en Espacio PCA 2D (Antes de Clustering)')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()


# VISUALIZACIÓN PCA 2D (BIPLOT)


# Calcular loadings para el biplot
loadings = pca.components_.T * np.sqrt(pca.explained_variance_)
loading_df = pd.DataFrame(loadings, columns=['PC1', 'PC2'], index=audio_features)

# Crear el biplot con un muestreo para mejor rendimiento
plt.figure(figsize=(12, 9))
sns.scatterplot(data=df_pca.sample(5000), x='PC1', y='PC2', alpha=0.5, s=30, color='lightgray', label='Canciones')

# Escala para las flechas
scale = 3
for i, feature in enumerate(audio_features):
    plt.arrow(0, 0, loading_df.loc[feature, 'PC1'] * scale, loading_df.loc[feature, 'PC2'] * scale,
              color='blue', alpha=0.8, width=0.01, head_width=0.3)
    plt.text(loading_df.loc[feature, 'PC1'] * scale * 1.1, 
             loading_df.loc[feature, 'PC2'] * scale * 1.1, 
             feature, color='darkred', ha='center', va='center', fontsize=10)

# Añadir ejes y etiquetas
plt.axhline(0, color='black', linewidth=0.8, alpha=0.5)
plt.axvline(0, color='black', linewidth=0.8, alpha=0.5)
plt.xlabel(f'PC1 ({pca.explained_variance_ratio_[0]:.1%})')
plt.ylabel(f'PC2 ({pca.explained_variance_ratio_[1]:.1%})')
plt.title('Biplot 2D: Influencia de las Características de Audio en PCA')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

```


* **Varianza explicada**: Las dos primeras componentes retienen ~39.2 % de la información del conjunto original. Esto indica que gran parte de la variabilidad está en dimensiones adicionales, pero PC1 y PC2 aportan un resumen significativo y fácilmente interpretable.
* **Nube PCA 2D**: La dispersión de puntos es amplia y continua; no hay agrupamientos claramente separados visualmente. Esto sugiere que los estilos musicales se solapan y que los límites entre “géneros acústicos” o “bailables” son difusos en las variables medidas.
* **Biplot — interpretación de flechas**:

  * **Flechas largas a la derecha**: `energy` y `loudness` tienen alta carga positiva en **PC1** → PC1 = *intensidad / producción sonora*.
  * **Flechas arriba-derecha**: `danceability` y `valence` cargan en **PC2** → PC2 = *bailabilidad / carácter emocional*.
  * **`acousticness`** apunta hacia la izquierda (opuesto a energy/loudness): evidencia la **anticorrelación** entre producción energética y carácter acústico.
  * **`instrumentalness`** se proyecta hacia la izquierda-abajo: las pistas instrumentales se diferencian de las canciones energéticas y bailables.
  * **Variables cercanas al origen** (`speechiness`, `key`, `mode`): contribuyen poco a PC1–PC2 y su efecto no se aprecia bien en este plano.
* **Relaciones entre variables**: Las direcciones relativas y ángulos de las flechas confirman correlaciones previas (ej., `energy` ≈ `loudness`; `energy` ≈ −`acousticness`; `danceability` ≈ `valence`).


## **Mejor K**

```python

# Metodo del codo 
wcss = []
k_values = range(1, 11)

for k in k_values:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(X_scaled)
    wcss.append(kmeans.inertia_)

plt.figure(figsize=(10, 6))
plt.plot(k_values, wcss, 'bo-')
plt.title('Método del Codo para Selección de K')
plt.xlabel('Número de Clústeres (K)')
plt.ylabel('suma de cuadrados dentro del clúster (WCSS)')
plt.grid(True)
plt.show()
```

El código implementó el algoritmo **K-Means** del paquete `sklearn.cluster`, con el objetivo de agrupar las canciones según sus características acústicas (como *energy, danceability, loudness, acousticness*, entre otras).

Los pasos principales fueron:

1. **Inicialización:**
   Se definió un rango de valores de K (por ejemplo, de 1 a 10) para evaluar distintos números de clusters.

2. **Cálculo del WCSS:**
   Para cada valor de K, se ejecutó el modelo K-Means y se registró su WCSS, que indica la compactación de los datos dentro de cada grupo.

3. **Visualización del Codo:**
   Finalmente, se graficó el número de clusters frente al WCSS, obteniendo la curva que permite identificar visualmente el punto óptimo (el “codo”).

4. **Selección de K óptimo:**
   Con base en la gráfica, se eligió **K = 4** para el modelo final, lo que permite una agrupación representativa y manejable de las canciones según sus rasgos musicales.

El gráfico muestra el resultado de aplicar el **método del codo (Elbow Method)** para determinar el número óptimo de *clusters* en el modelo **K-Means**.
En el eje X se representa el número de posibles grupos (K), y en el eje Y el **WCSS (Within-Cluster Sum of Squares)**, que mide la suma de las distancias cuadradas de los puntos respecto al centroide de su grupo.

Se observa que a medida que aumenta el número de *clusters*, el valor de WCSS disminuye, lo cual es esperado, ya que al incrementar los grupos, cada conjunto de puntos está más cerca de su centroide.
Sin embargo, esta reducción deja de ser significativa a partir de cierto punto, donde la curva se “dobla” formando un **codo**.

En este caso, el codo se aprecia alrededor de **K = 4**, lo que indica que dividir el conjunto de datos en **cuatro grupos** logra un equilibrio adecuado entre simplicidad del modelo y capacidad explicativa.
Por debajo de ese valor, los clusters no capturan toda la variabilidad de los datos, y por encima, se incrementa la complejidad sin una mejora sustancial en la cohesión interna.


## **Análisis Silhouetee**

```python

# análisis Silhouetee
silhouette_scores = []
k_values_sil = range(2, 11)

for k in k_values_sil:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    cluster_labels = kmeans.fit_predict(X_scaled)
    silhouette_avg = silhouette_score(X_scaled, cluster_labels)
    silhouette_scores.append(silhouette_avg)

plt.figure(figsize=(10, 6))
plt.plot(k_values_sil, silhouette_scores, 'ro-')
plt.title('valores Silhouette para Diferentes K')
plt.xlabel('Número de Clústeres (K)')
plt.ylabel('valores Silhouette')
plt.grid(True)
plt.show()

# k óptimo
optimal_k = k_values_sil[np.argmax(silhouette_scores)]
print(f"Número óptimo de clústers: {optimal_k}")
```


El código realiza un análisis de silueta de la siguiente manera:

1. **Configuración del rango de K:**
   Se evaluaron valores de K entre 2 y 10, aplicando el algoritmo **K-Means** en cada caso sobre los datos estandarizados (`X_scaled`).

2. **Cálculo del Silhouette Score:**
   Para cada K, se calcularon las etiquetas de los clusters (`fit_predict`) y se estimó el promedio del **Silhouette Score**, que mide la distancia media entre puntos del mismo grupo y de diferentes grupos.

3. **Visualización del comportamiento:**
   Los valores se graficaron para observar cómo cambia el puntaje al variar el número de clusters. El punto más alto indica la mejor estructura de agrupamiento desde el punto de vista geométrico.

4. **Determinación del K óptimo:**
   El código imprime el valor de K con el mayor puntaje de silueta (en este caso, **K = 2**).
   No obstante, al combinar este resultado con el análisis del método del codo y la interpretación del contexto, se estableció **K = 4 como la elección final** para desarrollar los clústeres definitivos.



El gráfico representa los valores del **coeficiente de silueta** para diferentes números de *clusters* (K), calculados con el algoritmo **K-Means**.
El **Silhouette Score** mide la calidad de la separación entre grupos:

* Un valor cercano a **1** indica que los puntos están bien asignados a su clúster y distantes de los demás.
* Un valor cercano a **0** sugiere superposición entre grupos.
* Valores negativos implican asignaciones incorrectas.

En este caso, el puntaje de silueta más alto se obtuvo para **K = 2**, con un valor aproximado de **0.21**. Esto indica que, desde un punto de vista puramente estadístico, el modelo con dos clústeres ofrece la mejor cohesión y separación entre los grupos.

Sin embargo, al analizar la naturaleza del conjunto de datos y los resultados de los otros métodos (como el **método del codo** y la **interpretación PCA**), se determinó que **K = 4** representa una solución más **informativa y práctica**.
Esta elección equilibra la simplicidad del modelo con una segmentación más detallada de las canciones, permitiendo identificar mejor las diferencias entre patrones musicales (por ejemplo, canciones energéticas, acústicas, bailables, etc.).

Por tanto, **aunque el análisis de silueta sugiere K = 2, la decisión final se mantuvo en K = 4**, respaldada por criterios interpretativos y de valor analítico para el estudio.


## **K means**

```python

optimal_k = 4  
kmeans = KMeans(n_clusters=optimal_k, random_state=42, n_init=10)
cluster_labels = kmeans.fit_predict(X_scaled)

# Add to df
df['cluster'] = cluster_labels
```

```python

plt.figure(figsize=(12, 8))

# Paleta colores para los clústeres
unique_clusters = np.unique(cluster_labels)
colors = plt.cm.tab10(np.linspace(0, 1, len(unique_clusters)))

# Dibujar los puntos por grupo
for cluster_id, color in zip(unique_clusters, colors):
    plt.scatter(
        X_pca[cluster_labels == cluster_id, 0],
        X_pca[cluster_labels == cluster_id, 1],
        color=color,
        alpha=0.6,
        label=f'Cluster {cluster_id + 1}'
    )

plt.title('visualización de Clústeres K-Means en Espacio PCA 2D')')
plt.xlabel(f'Primer componente (Varianza expliada: {pca.explained_variance_ratio_[0]:.2%})')
plt.ylabel(f'Segundo componente (Varianza expliada: {pca.explained_variance_ratio_[1]:.2%})')
plt.legend(title='Clusters')
plt.grid(True)
plt.show()
```


En esta sección se **aplica el algoritmo K-Means** utilizando el valor de **K = 4**, determinado previamente mediante el método del codo y el análisis de la silueta.
El objetivo es **agrupar los datos en cuatro conjuntos** que compartan características similares entre sí.

* `KMeans(n_clusters=optimal_k, random_state=42, n_init=10)`
  Crea el modelo de clustering.

  * `n_clusters=4` indica que se formarán 4 grupos.
  * `random_state=42` asegura que los resultados sean reproducibles.
  * `n_init=10` repite el proceso varias veces con diferentes centroides iniciales y conserva la mejor solución (la de menor error).

* `fit_predict(X_scaled)` realiza dos acciones:

  1. **Entrena** el modelo con los datos estandarizados (`fit`).
  2. **Predice** a qué grupo pertenece cada observación (`predict`).

* Finalmente, `df['cluster'] = cluster_labels` agrega una nueva columna al DataFrame original con la etiqueta de cluster asignada a cada registro.

 En otras palabras, **cada fila del dataset (por ejemplo, una canción)** queda clasificada dentro de uno de los cuatro clusters definidos por el modelo.



El gráfico final muestra **cuatro grupos diferenciados de observaciones**, cada uno identificado con un color distinto.
Aunque la separación entre ellos no es completamente rígida (algo común en datos multidimensionales como los musicales), se aprecia una **tendencia natural de agrupamiento**, donde las canciones con características acústicas similares se ubican más próximas en el plano.




## **Métricas de evaluación**

```python


# Silhouette final 
final_silhouette = silhouette_score(X_scaled, cluster_labels)
print(f"Final Silhouette Score for K={optimal_k}: {final_silhouette:.3f}")

# Métricas adicionales 
ch_score = calinski_harabasz_score(X_scaled, cluster_labels)
db_score = davies_bouldin_score(X_scaled, cluster_labels)

print(f"Calinski-Harabasz Score: {ch_score:.3f}")
print(f"Davies-Bouldin Score: {db_score:.3f}")
```


Para evaluar la calidad del modelo de agrupamiento se aplicaron tres métricas de validación interna: **Silhouette Score**, **Calinski-Harabasz Score** y **Davies-Bouldin Score**, obteniendo los siguientes resultados:

* **Silhouette Score:** 0.145
* **Calinski-Harabasz Score:** 16821.841
* **Davies-Bouldin Score:** 2.002

El **Silhouette Score** mide la coherencia interna de los clusters, es decir, qué tan cerca están los puntos dentro de un mismo grupo en comparación con otros. Un valor de **0.145** indica que los clusters presentan **una separación moderada**, aunque con cierto **solapamiento entre ellos**.

El **Calinski-Harabasz Score**, con un valor alto (**16821.841**), sugiere que **existe una separación general adecuada entre los grupos**, ya que los centroides se encuentran a distancias considerables unos de otros.

Por su parte, el **Davies-Bouldin Score**, con un valor de **2.002**, refuerza la idea de que **algunos clusters mantienen similitud entre sí**, lo que implica que las fronteras entre grupos no son completamente definidas.

En conjunto, estos resultados muestran que el modelo con **K = 4** genera una **estructura de agrupamiento aceptable**, donde se logra una diferenciación básica entre grupos, pero con **posibles mejoras en la cohesión interna**. Se recomienda contrastar con otros valores de *K* o aplicar métodos alternativos (como DBSCAN) para obtener una segmentación más precisa.


## **Sistema de recomendación**

```python

from sklearn.preprocessing import OneHotEncoder

# Preparar features con clusters
encoder = OneHotEncoder(sparse_output=False)
cluster_onehot = encoder.fit_transform(df[['cluster']])
X_final = np.hstack([X_pca, cluster_onehot])

# Matriz de similitud
cosine_sim = cosine_similarity(X_final)

# Función de recomendación
def recomendar_cancion(track_name, top_n=10):
    if track_name not in df['track_name'].values:
        return f"Canción '{track_name}' no encontrada."
    
    idx = df[df['track_name'] == track_name].index[0]
    sim_scores = sorted(list(enumerate(cosine_sim[idx])), key=lambda x: x[1], reverse=True)[1:top_n+1]
    indices = [i[0] for i in sim_scores]
    
    results = df.iloc[indices].copy()
    results['Similitud'] = [score[1] for score in sim_scores]
    return results[['track_name', 'artists', 'track_genre', 'cluster', 'Similitud']]

# Ejemplo

display(recomendar_cancion('Shape of You'))
```


En esta etapa se integró el modelo de *clustering* con un **sistema de recomendación basado en similitud de características**.
Primero, se aplicó **One-Hot Encoding** sobre las etiquetas de los clusters generados por K-Means, con el fin de convertirlos en variables numéricas que pudieran ser utilizadas en el cálculo de similitudes. Luego, estas variables se combinaron con los componentes principales obtenidos del PCA para formar la **matriz final de características (X_final)**.

Posteriormente, se calculó una **matriz de similitud coseno**, la cual mide el grado de similitud entre canciones según sus características acústicas y el cluster al que pertenecen.
Finalmente, se implementó una **función de recomendación** que, dado el nombre de una canción, busca los elementos más similares dentro del conjunto de datos y devuelve las canciones más parecidas, junto con su artista, género, cluster y nivel de similitud.

Este proceso permite ofrecer **recomendaciones personalizadas**, agrupando las canciones por patrones comunes y reforzando la coherencia entre las sugerencias del sistema.




El sistema de recomendación se probó con la canción **“Shape of You”** como punto de referencia. A partir de la matriz de similitud coseno, se identificaron las canciones con características más cercanas en el espacio de componentes principales y en los grupos formados por el modelo K-Means.

Los resultados muestran un **alto nivel de similitud (superior al 0.99)** entre la canción base y las canciones recomendadas, todas pertenecientes al **Cluster 3**. Este comportamiento indica que las canciones dentro de este grupo comparten **atributos acústicos semejantes**, tales como energía, ritmo, tonalidad o valencia emocional, lo cual respalda la coherencia del modelo de agrupamiento.

Además, la variedad de géneros observada (pop español, salsa, forró, dancehall y R&B) sugiere que el modelo no clasifica únicamente por género musical, sino por **afinidad sonora y patrón rítmico**, lo que amplía la capacidad del sistema para realizar recomendaciones más diversas pero perceptualmente consistentes.

En conclusión, el sistema logró **recomendar canciones con similitud estructural y acústica relevante**, demostrando que la integración del análisis de componentes principales (PCA), el modelo de clustering (K-Means) y la medida de similitud coseno ofrece un enfoque efectivo para la recomendación automatizada de música.



## **Visualización de recomendaciones**

```python



def plot_recomendaciones(track_name, top_n=5):
    if track_name not in df['track_name'].values:
        print("Canción no encontrada.")
        return
    
    # Índice de la canción base
    idx = df[df['track_name'] == track_name].index[0]
    
    # Canciones más similares
    sim_scores = sorted(list(enumerate(cosine_sim[idx])), key=lambda x: x[1], reverse=True)[1:top_n+1]
    indices = [i[0] for i in sim_scores]
    
    # Colores discretos por cluster
    unique_clusters = np.unique(cluster_labels)
    colors = plt.cm.tab10(np.linspace(0, 1, len(unique_clusters)))
    
    # Gráfico base de todos los clusters
    plt.figure(figsize=(12, 8))
    for cluster_id, color in zip(unique_clusters, colors):
        plt.scatter(
            X_pca[cluster_labels == cluster_id, 0],
            X_pca[cluster_labels == cluster_id, 1],
            color=color,
            alpha=0.4,
            label=f'Cluster {cluster_id + 1}'
        )
    
    # Canción base en rojo
    plt.scatter(X_pca[idx, 0], X_pca[idx, 1], c='red', s=200, edgecolor='black', label='Canción base')
    
    # Recomendaciones en azul
    plt.scatter(X_pca[indices, 0], X_pca[indices, 1], c='blue', s=120, edgecolor='black', label='Recomendadas')
    
    # Configuración visual
    plt.title(f"Recomendaciones para: '{track_name}' (K-Means + PCA)")
    plt.xlabel('Componente Principal 1 (PC1)')
    plt.ylabel('Componente Principal 2 (PC2)')
    plt.legend()
    plt.grid(True, linestyle='--', alpha=0.5)
    plt.tight_layout()
    plt.show()

# Ejemplo de uso
plot_recomendaciones('Shape of You')
```


La canción *“Shape of You”* (en rojo) y sus recomendaciones (en azul) se agrupan dentro del mismo clúster, evidenciando similitud acústica y coherencia en el modelo.
